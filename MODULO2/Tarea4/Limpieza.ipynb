```python
import pandas as pd 
import seaborn as sns 
import numpy as np 
import matplotlib.pyplot as plt 
import warnings 
warnings.filterwarnings("ignore") 
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
# Cargar el dataset 
df = pd.read_csv('bank.csv')
```

```python
# Verificar si hay valores faltantes y llenarlos con la media de cada columna numérica 
df.fillna(df.mean(), inplace=True)
```

```python
# Verificar y eliminar valores atípicos 
Q1 = df.quantile(0.25) 
Q3 = df.quantile(0.75) 
IQR = Q3 - Q1 
df = df[~((df < (Q1 - 1.5 * IQR)) | (df > (Q3 + 1.5 * IQR))).any(axis=1)]
```

```python
# Transformar columnas categóricas
categorical_cols = ['job', 'marital', 'education', 'default', 'housing', 'loan', 'contact', 'month', 'poutcome', 'day'] 
encoder = OneHotEncoder(handle_unknown='ignore', sparse=False) 
encoded_df = pd.DataFrame(encoder.fit_transform(df[categorical_cols]))
```

```python
# Añadir nombres de las nuevas columnas 
encoded_df.columns = encoder.get_feature_names(categorical_cols)
```

```python
# Eliminar columnas categóricas originales y añadir las transformadas 
df.drop(categorical_cols, axis=1, inplace=True)
df = pd.concat([df, encoded_df], axis=1)
```

```python
# Transformar y normalizar columnas numéricas 
numeric_cols = ['age', 'balance', 'duration', 'campaign', 'pdays', 'previous'] 
scaler = StandardScaler() 
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])
```

```python
# Guardar el dataset limpio y transformado en un nuevo archivo CSV 
df.to_csv('bank_marketing_transformed.csv', index=False)
```
